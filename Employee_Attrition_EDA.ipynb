```python
continuous_cols = [
    "Age",
    "DailyRate",
    "DistanceFromHome",
    "HourlyRate",
    "MonthlyIncome",
    "MonthlyRate",
    "NumCompaniesWorked",
    "PercentSalaryHike",
    "TotalWorkingYears",
    "TrainingTimesLastYear",
    "YearsAtCompany",
    "YearsInCurrentRole",
    "YearsSinceLastPromotion",
    "YearsWithCurrManager"
]

df[continuous_cols].boxplot(figsize=(18, 8), rot=90)
plt.title("Boxplots of Continuous Variables")
plt.show()
```

```python
for col in continuous_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    
    print(f"{col}: {len(outliers)} potential outliers")
```

### 2. Which departments have higher attrition?

```python
pd.crosstab(df["Department"], df["Attrition"])
```

```python
pd.crosstab(df["Department"], df["Attrition"], normalize="index") * 100
```

```python
department_attrition = (
    pd.crosstab(
        df["Department"],
        df["Attrition"],
        normalize="index"
    )["Yes"] * 100
)

department_attrition
```

```python
plt.figure(figsize=(8, 5))

department_attrition.plot(kind="bar")

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)

plt.show()
```

### 3. Does overtime relate to attrition?

```python
pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100
```

```python
overtime_attrition = (
    pd.crosstab(
        df["OverTime"],
        df["Attrition"],
        normalize="index"
    )["Yes"] * 100
)

plt.figure(figsize=(6, 4))

overtime_attrition.plot(kind="bar")

plt.title("Attrition Rate by Overtime")
plt.xlabel("OverTime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)

plt.show()
```

### 4.Does salary/income differ between employees who leave and stay?




```python
df.groupby("Attrition")["MonthlyIncome"].agg(
    ["count", "mean", "median", "min", "max"]
)
```

```python
plt.figure(figsize=(7, 5))

df.boxplot(column="MonthlyIncome", by="Attrition")

plt.title("Monthly Income by Attrition")
plt.suptitle("")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")

plt.show()
```

### 5. How does job satisfaction relate to attrition?

```python
pd.crosstab(
    df["JobSatisfaction"],
    df["Attrition"],
    normalize="index"
) * 100
```

```python
job_sat_attrition = pd.crosstab(
    df["JobSatisfaction"],
    df["Attrition"],
    normalize="index"
) * 100

job_sat_attrition.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Attrition Rate by Job Satisfaction")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)
plt.legend(title="Attrition")

plt.show()
```

### 6. Does distance from home have any relationship with attrition?

```python
df.groupby("Attrition")["DistanceFromHome"].agg(
    ["count", "mean", "median", "min", "max"]
)
```

```python
plt.figure(figsize=(7, 5))

df.boxplot(
    column="DistanceFromHome",
    by="Attrition"
)

plt.title("Distance From Home by Attrition")
plt.suptitle("")
plt.xlabel("Attrition")
plt.ylabel("Distance From Home")

plt.show()
```

```python
df["DistanceGroup"] = pd.cut(
    df["DistanceFromHome"],
    bins=[0, 5, 10, 20, 30],
    labels=["1-5", "6-10", "11-20", "21-30"]
)

pd.crosstab(
    df["DistanceGroup"],
    df["Attrition"],
    normalize="index"
) * 100
```

### 7. Which job roles have higher attrition?


```python
jobrole_attrition = (
    pd.crosstab(
        df["JobRole"],
        df["Attrition"],
        normalize="index"
    )["Yes"] * 100
)

jobrole_attrition.sort_values(ascending=False)
```

```python
plt.figure(figsize=(10, 6))

jobrole_attrition.sort_values().plot(kind="barh")

plt.title("Attrition Rate by Job Role")
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Job Role")
plt.show()
```

### 8. Does age influence attrition patterns?

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 60],
    labels=["18-25", "26-35", "36-45", "46-60"]
)
```

```python
age_attrition = (
    pd.crosstab(
        df["AgeGroup"],
        df["Attrition"],
        normalize="index"
    )["Yes"] * 100
)

age_attrition
```

```python
plt.figure(figsize=(8, 5))

age_attrition.plot(kind="bar")

plt.title("Attrition Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)

plt.show()
```

### 10. What differences can be observed between employees who left and those who stayed?


```python
comparison = df.groupby("Attrition")[
    [
        "Age",
        "MonthlyIncome",
        "DistanceFromHome",
        "JobSatisfaction",
        "WorkLifeBalance",
        "TotalWorkingYears",
        "YearsAtCompany"
    ]
].mean()

comparison
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 9))

variables = [
    "Age",
    "MonthlyIncome",
    "DistanceFromHome",
    "JobSatisfaction",
    "WorkLifeBalance",
    "YearsAtCompany"
]

for ax, variable in zip(axes.flat, variables):
    df.boxplot(
        column=variable,
        by="Attrition",
        ax=ax
    )
    ax.set_title(variable)
    ax.set_xlabel("Attrition")
    ax.set_ylabel(variable)

plt.suptitle("Comparison of Employees Who Stayed and Left")
plt.tight_layout()
plt.show()
```
